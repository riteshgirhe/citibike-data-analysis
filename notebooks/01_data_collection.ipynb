```python
import os
import requests
import pandas as pd

# 1. Create external folder
os.makedirs("../data/external", exist_ok=True)

# 2. Open-Meteo API URL
url = "https://archive-api.open-meteo.com/v1/archive"

# 3. API parameters
params = {
    "latitude": 40.71,
    "longitude": -74.00,
    "start_date": "2024-06-01",
    "end_date": "2024-08-31",
    "hourly": "temperature_2m,relative_humidity_2m,precipitation,wind_speed_10m",
    "timezone": "America/New_York"
}

# 4. Send request to API
response = requests.get(url, params=params)

# 5. Check whether request was successful
print("Status code:", response.status_code)

# 6. Convert response to JSON
data = response.json()

# 7. Convert hourly data into DataFrame
weather_df = pd.DataFrame(data["hourly"])

# 8. Convert time column to datetime
weather_df["time"] = pd.to_datetime(weather_df["time"])

# 9. Display first 5 rows
print(weather_df.head())

# 10. Display number of rows and columns
print("Shape:", weather_df.shape)

# 11. Save weather data
weather_df.to_csv(
    "../data/external/nyc_weather.csv",
    index=False
)

print("Weather data saved successfully!")
```
